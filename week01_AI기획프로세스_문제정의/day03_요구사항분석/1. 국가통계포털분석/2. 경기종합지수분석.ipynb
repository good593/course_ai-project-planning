## 상관관계 분석

```python
corr = df.corr()
corr
```

```python
corr.columns
```

```python
corr["재고순환지표(전월차)(%p)"].nsmallest(10)
```

```python
mask = np.triu(np.ones_like(corr))

plt.figure(figsize=(15, 15))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1, mask=mask)
```
